# Strings (Concatenation, Slicing and Other methods) :

# 1. String Concatenation: 

```python
String1="Hello "
String1
```

```python
String2=input("Enter your name: ")
String2
concat_string=String1+String2
concat_string
```

```python
print(concat_string)
```

```python
String3=", welcome to Python programming"
String3
```

```python
Concat=concat_string+String3
Concat
```

```python
print(Concat)
```

# 2. String Slicing and Indexing: 

```python
print("first character:",Concat[0])
```

```python
print("last character:",Concat[-1])
```

```python
print("first 5 characters:",Concat[:5])
```

```python
print("last 11 characters:",Concat[-11:])
```

```python
print("string in reverse:",Concat[::-1])
```

```python
start = Concat.find("Python")
end = start + len("Python")
print(Concat[start:end])
```
